Q1 Stop sending emails after 3 users (Using for loop and break)

users = ["Ali", "Sara", "John", "Zara", "Mike"]

Output:

Email sent to Ali

Email sent to Sara

Email sent to John

Limit reached. Stop sending emails.

```python
users = ["Ali", "Sara", "John", "Zara", "Mike"]
a=0
for i in users:
    a = a+1
    print(f"Email sent to {i}")
    if a==3:
        print("Limit reached. Stop sending emails.")
        break
```

Q2. Stop monitoring temperature if critical temperature reached (Using for loop and break)

temps = [22, 25, 30, 45, 28] critical = 40

Ouput:

Temperature: 22

Temperature: 25

Temperature: 30

Critical temperature reached! Stop monitoring.

```python
temps = [22, 25, 30, 45, 28]
critical = 40
for i in temps:
    if i < critical:
        print(f"Temperature: {i}")
    else:
        print("Critical temperature reached! Stop monitoring.")
        break
```

Q3. Calculate total price of items in a shopping cart (Using For loop)

cart = [100, 250, 50, 300]

Output:

Total Price: 700

```python
cart = [100, 250, 50, 300]
total_price = 0
for price in cart:
    total_price += price

print(f"Total Price: {total_price}")
```

Q8. Create a multiplication table using a nested for loop.

numbers = [2, 3, 4]
multipliers = [1, 2, 3]

Expected Output:

2 x 1 = 2
2 x 2 = 4
2 x 3 = 6
3 x 1 = 3
3 x 2 = 6
3 x 3 = 9
4 x 1 = 4
4 x 2 = 8
4 x 3 = 12

Use a nested for loop.

```python
numbers = [2, 3, 4]
multipliers = [1, 2, 3]

for num in numbers:
    for mul in multipliers:
        a = num * mul

        print(f"{num} x {mul} = {a}")

```

Q9. Keep asking the user for a password until the correct password is entered.

correct_password = "python123"

Expected behaviour:

Enter password: abc
Incorrect password. Try again.

Enter password: 12345
Incorrect password. Try again.

Enter password: python123
Access granted!

Use a while loop.

```python
correct_password = "python123"

while True:
    passw = input("Enter password: ")

    if passw == correct_password:
        print(f"Access granted !")
        break
       
    else:
        print("Incorrect password. Try again.")
```

Q10. Count down from 5 to 1 and then display "Time's up!".

Expected Output:

5
4
3
2
1
Time's up!

Use a while loop.

```python
numbers = list(range(1,6))
lst = numbers[::-1]

for i in lst:
    print(i, end=" ")

print("Times up !")
```
