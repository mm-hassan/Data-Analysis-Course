```python
a = int(input("Enter English Marks: "))
if a>100:
    print("Not possible greater than 100. Try Again")
else:   
    b = int(input("Enter Urdu Marks: "))
    
    if b>100:
        print("Not possible greater than 100. Try Again")
    else: 
        c = int(input("Enter Maths Marks: "))
       
        if c>100:
            print("Not possible greater than 100. Try Again")
        else: 
            d = int(input("Enter Physics Marks: "))
        
        if d>100:
            print("Not possible greater than 100. Try Again")
        else: 
            e = int(input("Enter Chemistry Marks: "))

            Total_Obt = a+b+c+d+e
            percentage = float((Total_Obt/500 *100))

            print ("-----------------------------------------")
            print("                RESULT                    ")
            print ("-----------------------------------------")
            print("Marks Obtained", Total_Obt)
            print("Your Percentage", percentage)
            
            if percentage >= 80 and percentage <= 100:
                print("Grade A")
            
            elif percentage >= 60 and percentage <= 80:
                print("Grade B")
            
            elif percentage >= 50 and percentage <= 60:
                print("Grade C")
            
            else: 
                print("Fail")
```
