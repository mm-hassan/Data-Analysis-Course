Q1. Write a program that takes two numbers (a = 15 , b = 4) and prints their:


- Sum

- Difference

- Product

- Division

- Floor Division

- Modulus

```python
a= 15
b= 4
summ = a+b
dif = a-b
pro = a*b
div = a/b
fdiv = a//b
mod = a%b
print("Sum is: ", summ)
print("Difference is: ", dif)
print("Product is: ", pro)
print("Division is: ", div)
print("Floor Division is: ", fdiv)
print("Modulus is: ", mod)
```

Q14.
text = "    Data Analysis with Python is very powerful    "

Perform the following tasks:

Remove extra spaces using strip().
Convert the string into title case using title().
Replace "Powerful" with "Important".
Find the position of the word "Python".

```python
text = "    Data Analysis with Python is very powerful    "
Rem_Spaces = text.strip()

Title_Case = Rem_Spaces.title()

Replace = Title_Case.replace("Powerful","Important")

Find_Pos = text.find("Python")

print(Rem_Spaces)
print(Title_Case)
print(Replace)
print(Find_Pos)
```
