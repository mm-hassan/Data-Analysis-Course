Q2: Age Group Classification 

Write a Python program that takes the age of a person as input and classifies them into one of the following age groups:

"Child" if the age is between 0 and 12 (inclusive).

"Teenager" if the age is between 13 and 19 (inclusive).

"Adult" if the age is between 20 and 64 (inclusive).

"Senior" if the age is 65 or older.


```python
age = int(input("Enter your age: "))
if age>= 0 and age<= 12:
    print("Child")
elif age>= 13 and age<= 19:
    print("Teenager")
elif age>= 20 and age<= 64:
    print("Adult")
else:
    print("Senior")
```

Q8. inventory = { &quot;The Great Gatsby&quot;: {&quot;price&quot;: 10.99, &quot;quantity&quot;: 5}, &quot;Rich dad poor dad&quot;: {&quot;price&quot;:
8.99, &quot;quantity&quot;: 8}, &quot;Harry Potter&quot;: {&quot;price&quot;: 12.50, &quot;quantity&quot;: 2}, }
• Add a new book in inventory using method and get all book titles using method
• Remove a book if it&#39;s out of stock

```python
inventory = {
    "The Great Gatsby": {"price": 10.99, "quantity": 5},
    "Rich dad poor dad": {"price": 8.99, "quantity": 8},
    "Harry Potter": {"price": 12.50, "quantity": 2}
}

inventory.update({"Atomic Habits": {"price": 15.00, "quantity": 7}}) 

book_titles = inventory.keys()
print("All Book Titles:", book_titles)

for book in list(inventory.keys()):
    if inventory[book]["quantity"] == 0:
        inventory.pop(book)
        print(f"Removed out-of-stock book: '{book}'")

print("Final Inventory:", inventory)
```

Q10. Write a Python program to check whether the entered letter is a vowel, a consonant, or invalid input.

The program should take a single character as input. If the character is a vowel (a, e, i, o, u in both upper and lower case), it will print "Vowel". If it is an alphabet but not a vowel, it will print "Consonant". Otherwise, it will print "Please enter valid alphabet".

Expected Output:

Enter a single letter:  a

a is a Vowel

---

Enter a single letter:  E

E is a Vowel

---

Enter a single letter:  z

z is a Consonant

---

Enter a single letter:  7

Please enter valid alphabet


```python
alpha = input("Please Enter a valid alphabet: ")
if alpha.isalpha() and len(alpha) == 1:
    if alpha.lower() in ('a','e','i','o','u'):
        print(f"{alpha} is Vowel")
    else:
        print(f"{alpha} is a Consonant")
else:
    print("Invalid alphabet")
```
